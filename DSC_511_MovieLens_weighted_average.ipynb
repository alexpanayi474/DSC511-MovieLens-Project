## EDA Part 1: Movies Released Per Year

We begin by extracting the **release year** from each movie title using a regular expression,
since the year is embedded in the title string (e.g. *"Toy Story (1995)"*).

This analysis is performed on the **full `movies_df`** dataset (86,537 movies), which includes
movies that were never rated on MovieLens. This gives a more accurate picture of movie
production over time, compared to using only the joined DataFrame which would exclude ~3,300
unrated movies.

> **Note:** This chart reflects movies *catalogued on MovieLens*, not the entire global film
> production — so it carries a bias toward popular and English-language films.

```python
# Extract year from title string using regex, then groupBy year
movies_with_year = movies_df.withColumn("release_year",regexp_extract(col("title"), r"\((\d{4})\)", 1))

movies_per_year = (movies_with_year
    .filter(col("release_year") != "")     # remove empty matches BEFORE casting
    .withColumn("release_year", col("release_year").cast("int")) # regexp_extract always returns a string (e.g. "1995"), .cast() converts a column from one data type to another
    .groupBy("release_year").agg(count("movieId").alias("num_movies")).orderBy("release_year").toPandas()
)

plt.figure(figsize=(14, 5))
plt.bar(movies_per_year["release_year"], movies_per_year["num_movies"], color="steelblue", width=0.8)
plt.title("Number of Movies (in MovieLens Website) Released Per Year ", fontsize=14)
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()
```

## Part 2: MovieLens Platform Popularity Over Time
To understand how the **MovieLens platform grew over time**, we analyze the total number
of ratings submitted by users in each **6-month period** from 1997 (platform launch) to 2023.

We use the `ratings_clean` DataFrame which already has the `ts` column — a proper Spark
timestamp converted from the raw Unix epoch integer in `ratings_df`.

### Why 6-month bins?
Monthly bins would produce ~300 bars (too noisy). Yearly bins would hide seasonal patterns.
6-month bins give a clean balance — enough granularity to see trends without clutter.


```python
# Count total ratings per 6-month period from 1997 to 2023
# Group by 6-month periods to track platform growth.
from pyspark.sql.functions import year, month, concat_ws, lpad, count, col, when,floor

ratings_timeline = (ratings_clean

    # Step 1: Extract year and month from the timestamp column
    .withColumn("year",  year(col("ts")))
    .withColumn("month", month(col("ts")))

    # floor(month / 6) == 0 → months 1-5 → H1
    # floor(month / 6) != 0 → months 6-12 → H2
    .withColumn("half",
        when(floor(col("month") / 6) == 0, "H1").otherwise("H2")
    )

    # Step 3: Build readable period label e.g. "2005-H1", "2005-H2"
    .withColumn("period_label",
        concat_ws("-", col("year").cast("string"), col("half"))
    )

    # Step 4: Keep only ratings from 1997 (platform launch) to 2023
    #.filter((col("year") >= 1997) & (col("year") <= 2023))

    # Step 5: Group by period and count total ratings in each 6-month bin
    .groupBy("year", "half", "period_label")
    .agg(count("rating").alias("total_ratings"))

    # Step 6: Sort chronologically — H1 before H2 within each year
    .orderBy("year", "half")

    .toPandas()
)

plt.figure(figsize=(18, 5))
plt.bar(range(len(ratings_timeline)), ratings_timeline["total_ratings"], color="coral")
plt.xticks(
    range(0, len(ratings_timeline), 2),       # show every 2nd label to avoid overlap
    ratings_timeline["period_label"].iloc[::2],
    rotation=45, ha="right", fontsize=8
)
plt.title("MovieLens Platform Activity — Total Ratings per 6-Month Period (1995–2023)", fontsize=13)
plt.xlabel("Period (6-month bins)")
plt.ylabel("Total Ratings")
plt.tight_layout()
plt.show()
```

### Observations

- **1996-H2 (highest peak ~1.45M)**: Surprisingly the most active period — likely a
  large batch import of retrospective ratings when the platform launched, where early
  adopters rated many movies at once.
- **1999-H2 → 2001-H2**: Second surge coinciding with the **DVD boom** and growing
  internet adoption.
- **2005-H1**: Notable peak likely driven by **Netflix's growing popularity** pushing
  users to discover and rate more films.
- **2007–2014 (steady decline)**: Gradual drop as competing platforms
  (IMDb, Rotten Tomatoes) fragmented the user base.
- **2016-H1 (strong resurgence ~1.38M)**: The **streaming era** reignited interest
  in movie rating and discovery tools.
- **2023-H2 (sharp drop)**: Incomplete period — dataset collection cutoff,
  not a real decline in activity.

### Favourite Genres of the top 10 users

```python
from pyspark.sql.functions import explode, split

# Get the list of top 10 user IDs
top10_ids = [row["userId"] for row in top10_users.collect()]

# Filter merged df for only top 10 users, explode genres to get one per row,
# then count how many times each user rated each genre
top10_genres = movies_ratings_merged.filter(col("userId").isin(top10_ids)) \
    .withColumn("genre", explode(split(col("genres"), "\\|"))) \
    .groupBy("userId", "genre") \
    .agg(count("rating").alias("genre_count")) \
    .orderBy("userId", "genre_count", ascending=[True, False])

# For each user keep only their top 3 favorite genres
from pyspark.sql.window import Window
from pyspark.sql.functions import rank

window = Window.partitionBy("userId").orderBy(col("genre_count").desc())

top10_fav_genres = top10_genres.withColumn("rank", rank().over(window)) \
                               .filter(col("rank") <= 3)

top10_fav_genres.show(30, truncate=False)
```

#### Drama is the #1 favorite genre for almost all top 10 users, which aligns with it being the most common genre in the dataset overall. Comedy and Thriller consistently appear as the 2nd and 3rd favorite genres, reflecting their large presence in the MovieLens catalog. No user has Action or Sci-Fi in their top 3, despite these being popular genres, suggesting that highly active users prefer character-driven over action-driven films.

## Number of Ratings vs Average Rating per Movie

```python
# Aggregate number of ratings and average rating per movie
movie_stats = movies_ratings_merged \
    .groupBy("movieId", "title") \
    .agg(count("rating").alias("num_ratings"),
         round(avg("rating"), 2).alias("avg_rating"))

movie_stats_pd = movie_stats.toPandas()

plt.figure(figsize=(12, 6))
plt.scatter(movie_stats_pd["num_ratings"],
            movie_stats_pd["avg_rating"],
            alpha=0.3, s=5, color="steelblue")

plt.title("Number of Ratings vs Average Rating per Movie")
plt.xlabel("Number of Ratings")
plt.ylabel("Average Rating")
plt.tight_layout()
plt.show()

```

#### Movies with very few ratings (left side of the plot) show extreme average ratings, spanning the full range from 0.5 to 5.0. With only a handful of ratings, there is not enough data to get a reliable estimate of the true quality - the average is highly sensitive to each individual opinion.
#### As the number of ratings increases, the average rating converges towards 3.5-4.0. This is known as the "regression to the mean" effect: with more ratings, the extreme values balance each other out and the average stabilizes towards the true quality of the movie.
#### The most rated movie (~122,000 ratings) maintains an average above 4.0, confirming that high popularity and high quality can go hand in hand.
#### We can conclude that the number of ratings is a key factor in the reliability of the average. A high avg_rating is only meaningful when supported by a large number of ratings.

```python
# number of distinct movies per decade from 1870 to 1960
movies_ratings_merged \
    .withColumn("release_year", regexp_extract(col("title"), r"\((\d{4})\)", 1)) \
    .filter(col("release_year") != "") \
    .withColumn("release_year", col("release_year").cast("int")) \
    .withColumn("decade", (floor(col("release_year") / 10) * 10).cast("int")) \
    .filter((col("decade") >= 1870) & (col("decade") < 1980)) \
    .select("movieId", "decade") \
    .distinct() \
    .groupBy("decade") \
    .count() \
    .orderBy("decade") \
    .show()
```

#### How ratings of older movies (starting from the 30s up to the 80s) change over time

```python
# Rating Stability Analysis extended to 1930s-1980s
# We start from the 1930s because earlier decades (1870s-1920s) have too few movies (under 700) to produce statistically reliable average ratings.
# The 1930s have 2,260 movies which is a sufficient sample size.

rating_era = movies_ratings_merged \
    .withColumn("release_year", regexp_extract(col("title"), r"\((\d{4})\)", 1)) \
    .filter(col("release_year") != "") \
    .withColumn("release_year", col("release_year").cast("int")) \
    .withColumn("release_decade", (floor(col("release_year") / 10) * 10).cast("int")) \
    .filter((col("release_decade") >= 1930) & (col("release_decade") <= 1980)) \
    .withColumn("rating_year", year(to_timestamp(col("ts")))) \
    .withColumn("rating_decade", (floor(col("rating_year") / 10) * 10).cast("int")) \
    .groupBy("release_decade", "rating_decade") \
    .agg(round(avg("rating"), 2).alias("avg_rating"),
         count("rating").alias("num_ratings")) \
    .orderBy("release_decade", "rating_decade")

rating_era.show(50)
```

####  Observations:
##### This analysis tracks how movies from the 1930s-1980s are rated across different  decades, to investigate whether older movies are consistently well-rated over time or whether their ratings change as newer users discover them.
###### Movies from the 1930s received their highest ratings in the 1990s (3.98), declining steadily to 3.54 in the 2020s. The num_ratings drops sharply from 123,506 in the 2000s to only 20,062 in the 2020s — very few users engage with 1930s movies today.
##### Movies from the 1940s show the highest avg rating overall, peaking at 4.05 in the 1990s and declining to 3.70 in the 2020s. This suggests that 1940s classics (e.g. Casablanca, Citizen Kane) are highly regarded but increasingly rated by fewer users (33,126 in the 2020s vs 193,805 in the 2000s).
##### Movies from the 1950s follow the same pattern: highest in the 1990s (3.98), remaining relatively stable through the decades, reaching 3.79 in the 2020s. Notably their num_ratings stays more consistent than the 1930s-1940s.

##### Movies from the 1960s received their highest rating in the 1990s (3.91),  declining to 3.74 in the 2020s. The num_ratings drops from 479,683 in the 2000s to only 81,821 in the 2020s.
##### Movies from the 1970s follow a similar pattern: highest in the 1990s (3.86), declining to 3.78 in the 2020s. The 2000s had the most ratings (814,328),  suggesting that 1970s movies were most actively watched and rated in the 2000s.
##### Movies from the 1980s show a different trend: a dip in the 2000s (3.51) but a recovery in the 2010s (3.65) and 2020s (3.71). This is the only decade where ratings increase over time — possibly driven by nostalgia from users who grew up in the 1980s and are now rating those movies fondly.
##### Overall pattern across all decades: ratings are highest when given in the 1990s and decline gradually in later decades. This consistent decline suggests that newer users tend to judge older films more critically than earlier users did, partially explaining the high avg ratings of older decades seen in previous analyses.

```python
from pyspark.sql.functions import floor, regexp_extract

# Extract decade from release year (e.g. 1995 -> 1990s)
# Use try_cast instead of cast to handle empty strings gracefully (returns null instead of error)
genre_decade = movies_ratings_merged \
    .withColumn("release_year", regexp_extract(col("title"), r"\((\d{4})\)", 1)) \
    .filter(col("release_year") != "") \
    .withColumn("release_year", col("release_year").cast("int")) \
    .withColumn("decade", (floor(col("release_year") / 10) * 10).cast("int")) \
    .filter((col("decade") >= 1930) & (col("decade") <= 2020)) \
    .withColumn("genre", explode(split(col("genres"), "\\|"))) \
    .filter(col("genre") != "(no genres listed)") \
    .groupBy("decade", "genre") \
    .agg(round(avg("rating"), 2).alias("avg_rating"))

# Convert to pandas and pivot for heatmap format
genre_decade_pd = genre_decade.toPandas()
pivot = genre_decade_pd.pivot(index="genre", columns="decade", values="avg_rating")

import seaborn as sns

plt.figure(figsize=(14, 8))
sns.heatmap(pivot, annot=True, fmt=".2f", cmap="YlOrRd", linewidths=0.5)
plt.title("Average Rating per Genre per Decade")
plt.xlabel("Decade")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```

### Observations:
#### Film-Noir consistently receives the highest ratings across all decades,peaking at 4.11 in the 1970s. Notably it maintains above 4.00 from the 1930s through the 1970s — confirming it is a niche but timeless genre.
#### Sci-Fi shows an unusual dip to 2.74 in the 1940s — likely because very   few Sci-Fi movies existed then and they were poorly received compared to the golden era of Sci-Fi in the 1950s-1970s.

### The Problem: Inconsistent Release Years

MovieLens encodes the release year directly inside each movie's title (e.g., *Toy Story (1995)*). After fetching official release dates from TMDB via their API, we compared the year embedded in each title against the TMDB release date. A significant number of movies showed a **mismatch** between the two sources — meaning we could not be certain which year was correct.

### Attempted Fix: Cross-referencing with IMDb (via OMDb API)

For the **5,643 mismatched movies**, we made a second attempt to resolve the correct release year by querying the **OMDb API** (which surfaces IMDb data). The outcome was:

| Result | Count |
|---|---|
| IMDb date found & year matches title  | 1,626 → **fixed** |
| IMDb date found but year still mismatches  | 4,014 → set to `null` |
| IMDb returned no result  | 306 → set to `null` |
| Movies with no year in title (unverifiable)  | ~660 → set to `null` |

### Decision: Nullify Unresolvable Dates

When neither TMDB nor IMDb agreed with the year in the MovieLens title, we had **no reliable ground truth** to fall back on. Rather than propagating a potentially incorrect release year into our models we set the `release_date` to `null` for all unresolvable cases.

# Furhter Data Cleaning and Feature Engineering

## Feature Engineering: Cast & Director Frequency (Historical Popularity)
#### For each movie released on or after the date of the earliest rating in our dataset, we compute two new features:
#### 1. cast_frequency: For each actor in a movie's top cast, we count how many times that actor appeared in other movies released BEFORE the target movie (release_year - 1 or earlier). These counts are then summed across all cast members of the movie. This captures the cumulative historical popularity/experience of the cast at the time of the movie's release.
#### 2. director_frequency: Same logic applied to the movie's director(s). We count how many movies each director had directed before the target movie's release year and sum across all directors of the movie.
#### Important:
####    - The cutoff date is derived from the earliest rating timestamp in the dataset (after filtering ratings to only include movies present in the cleaned movies_enriched_df).
####    - The historical counts use the FULL dataset (including pre-cutoff movies) to ensure accurate frequency counts for early releases.
####    - We use release_year - 1 (strictly before) to avoid data leakage, i.e. we never count appearances from the same year as the target movie.
####    - For movies with multiple cast members / directors, individual counts are summed to produce a single aggregated feature per movie.

## Imputation & Merging of Cast & Director Frequency Features

### Distribution Analysis
Both cast_frequency and director_frequency exhibit heavily right-skewed distributions:
- **Cast Frequency**: The vast majority of movies have a cast frequency below 50, with a long tail extending past 250. This is expected — most movies feature relatively unknown or emerging actors, while a small number of blockbusters star highly prolific cast members.
- **Director Frequency**: Even more concentrated near zero, with most movies having a director frequency below 20 and a tail reaching ~100. This reflects the fact that most directors have directed only a handful of films at the time of any given release.

### Imputation Strategy
Not all movies in `movies_after_cutoff` have cast or director information available (357 missing top_cast, 1,928 missing director). After the join, these movies will have NULL cast_frequency and/or director_frequency values.

We impute missing values using the **median** rather than the mean because:
1. Both distributions are heavily right-skewed — the mean is inflated by a small number of high-frequency outliers and would overestimate the typical movie's cast/director popularity.
2. The median better represents the "typical" movie in the dataset.

### Time-Aware Imputation
To avoid data leakage, we compute the median **only from movies released strictly before the target movie's release date**. This ensures that each imputed value reflects only information that was available at the time of the movie's release — consistent with the logic used to compute the features themselves.

```python
# Join both frequency features to movies_after_cutoff
movies_with_features = movies_after_cutoff \
    .join(cast_frequency.select("movieId", "cast_frequency"), on="movieId", how="left") \
    .join(director_frequency.select("movieId", "director_frequency"), on="movieId", how="left")

# Check nulls before imputation
print("=== Nulls BEFORE imputation ===")
movies_with_features.select(
    count(when(col("cast_frequency").isNull(), 1)).alias("null_cast_freq"),
    count(when(col("director_frequency").isNull(), 1)).alias("null_dir_freq")
).show()

# Time-aware median imputation using self-join:
# Spark's Window functions don't support percentile/median natively,
# so we use a self-join to collect all non-null values from movies
# released strictly before the target movie, then compute the median.

# Step 1: collect non-null values with their dates
cast_non_null = movies_with_features \
    .filter(col("cast_frequency").isNotNull()) \
    .select("movieId", "release_date_parsed", "cast_frequency")

dir_non_null = movies_with_features \
    .filter(col("director_frequency").isNotNull()) \
    .select("movieId", "release_date_parsed", "director_frequency")

# Step 2: for each null movie, compute median from movies released before it
cast_nulls = movies_with_features.filter(col("cast_frequency").isNull()) \
    .select("movieId", "release_date_parsed")

cast_imputed = cast_nulls.alias("a").join(
    cast_non_null.alias("b"),
    col("b.release_date_parsed") < col("a.release_date_parsed"),
    "left"
).groupBy("a.movieId") \
 .agg(percentile_approx("b.cast_frequency", 0.5).alias("cast_frequency_imputed"))

dir_nulls = movies_with_features.filter(col("director_frequency").isNull()) \
    .select("movieId", "release_date_parsed")

dir_imputed = dir_nulls.alias("a").join(
    dir_non_null.alias("b"),
    col("b.release_date_parsed") < col("a.release_date_parsed"),
    "left"
).groupBy("a.movieId") \
 .agg(percentile_approx("b.director_frequency", 0.5).alias("director_frequency_imputed"))

# Step 3: merge imputed values back
movies_with_features = movies_with_features \
    .join(cast_imputed, on="movieId", how="left") \
    .join(dir_imputed, on="movieId", how="left") \
    .withColumn("cast_frequency",
        coalesce(col("cast_frequency"), col("cast_frequency_imputed"))) \
    .withColumn("director_frequency",
        coalesce(col("director_frequency"), col("director_frequency_imputed"))) \
    .drop("cast_frequency_imputed", "director_frequency_imputed")

# Verify no nulls remain
print("=== Nulls AFTER imputation ===")
movies_with_features.select(
    count(when(col("cast_frequency").isNull(), 1)).alias("null_cast_freq"),
    count(when(col("director_frequency").isNull(), 1)).alias("null_dir_freq")
).show()

print(f"Total movies with features: {movies_with_features.count()}")
```

## Runtime Imputation

The runtime variable has 510 missing values in the post-cutoff dataset (51,323 movies). The distribution is right-skewed with a median of 94 minutes (Q1=85, Q3=106), and ranges from 1 to 840 minutes. The extreme outlier (Cinématon, 12,480 min) observed in the full dataset was released before the cutoff date and is no longer present.

We impute missing runtime values using the **median** rather than the mean, since the distribution is right-skewed — the mean (~95 min) is pulled upward by long documentaries and epic films, while the median (94 min) better represents the typical movie duration.

As with cast and director frequency, the imputation is **time-aware**: for each movie with a missing runtime, we compute the median from all movies in the post-cutoff dataset that have a known runtime and were released **strictly before** the target movie's release date. This avoids data leakage by ensuring no future information is used during imputation.


```python
# Runtime imputation — median from movies released strictly before the target movie

# Step 1: separate non-null and null runtime movies
runtime_non_null = movies_with_features \
    .filter(col("runtime").isNotNull()) \
    .select("movieId", "release_date_parsed", "runtime")

runtime_nulls = movies_with_features \
    .filter(col("runtime").isNull()) \
    .select("movieId", "release_date_parsed")

print(f"Movies with runtime: {runtime_non_null.count()}")
print(f"Movies missing runtime: {runtime_nulls.count()}")

# Step 2: for each null movie, compute median runtime from movies released before it
runtime_imputed = runtime_nulls.alias("a").join(
    runtime_non_null.alias("b"),
    col("b.release_date_parsed") < col("a.release_date_parsed"),
    "left"
).groupBy("a.movieId") \
 .agg(percentile_approx("b.runtime", 0.5).alias("runtime_imputed"))

# Step 3: merge back
movies_with_features = movies_with_features \
    .join(runtime_imputed, on="movieId", how="left") \
    .withColumn("runtime",
        coalesce(col("runtime"), col("runtime_imputed"))) \
    .drop("runtime_imputed")

# Verify
print(f"\nNull runtime after imputation: {movies_with_features.filter(col('runtime').isNull()).count()}")
```

## Rating Window Analysis: Conclusion

We attempted to define a **fixed post-release time window** (12, 18, or 24 months) to compute each movie's average rating as our prediction target. The idea was to capture "early audience reception" — ratings submitted shortly after a movie's release.

### Why This Approach Doesn't Work for Our Dataset

The histograms reveal a fundamental problem: **the vast majority of movies receive zero ratings within any reasonable post-release window**. Specifically:

- **12-month window**: Median coverage = 0.0%, meaning over half the eligible movies got no ratings at all in their first year. Only 1,441 movies excluded for being too recent.
- **18-month window**: Median coverage remains 0.0%, with 2,303 movies excluded.
- **24-month window**: The pattern continues — most movies still have 0% of their ratings in the first two years, and the number of excluded movies keeps growing.

This is because MovieLens is a **retrospective rating platform** — users rate movies they've watched at any point in their lives, not necessarily when a movie is first released. Unlike box office data or opening-weekend reviews, MovieLens ratings accumulate gradually over decades. A movie released in 2005 might receive most of its ratings between 2010 and 2023 as users discover it later.

Extending the window further (e.g., 5 or 10 years) would capture more ratings but would also **exclude an increasingly large portion of recent movies** from the dataset, defeating the purpose.

### Our Approach: Lifetime Average Rating After Release

Instead of a fixed time window, we compute each movie's **average user rating across all post-release ratings** up to the last rating date in the dataset (2023-07-20). This ensures:
1. Every movie in our dataset has a valid target variable.
2. We don't lose additional movies beyond the 1,013 already removed for having zero post-release ratings.
3. The target reflects the movie's overall audience reception on the platform.

```python
# Compute average post-release rating per movie as our target variable
avg_ratings = ratings_with_release \
    .groupBy("movieId") \
    .agg(
        round(avg("rating"), 4).alias("avg_rating"),
        count("*").alias("num_ratings")
    )

# Join to movies_with_features
movies_with_features = movies_with_features \
    .join(avg_ratings, on="movieId", how="inner")

print(f"Movies with features and target: {movies_with_features.count():,}")
#movies_with_features.show(10, truncate=False)
```

## Weighted Average Rating Target

The final CSV already contains `avg_rating` and `num_ratings`. A simple average rating can be noisy when a movie has very few ratings, so we create a Bayesian weighted average target. This keeps low-count movies in the data but pulls their ratings closer to the global mean.

Formula:

`weighted_avg_rating = (v / (v + m)) * R + (m / (v + m)) * C`

Where:

- `R` = movie average rating
- `v` = number of ratings for that movie
- `C` = global average rating across all ratings
- `m` = smoothing strength



```python
from pyspark.sql.functions import col, lit, sum as spark_sum, min, max, percentile_approx, count

# Check how many ratings exist per movie in the final modelling file
movies_with_features_df.select(
    count("*").alias("num_movies"),
    min("num_ratings").alias("min_num_ratings"),
    percentile_approx("num_ratings", 0.25).alias("q1_num_ratings"),
    percentile_approx("num_ratings", 0.50).alias("median_num_ratings"),
    percentile_approx("num_ratings", 0.75).alias("q3_num_ratings"),
    percentile_approx("num_ratings", 0.90).alias("p90_num_ratings"),
    percentile_approx("num_ratings", 0.95).alias("p95_num_ratings"),
    max("num_ratings").alias("max_num_ratings")
).show(truncate=False)

# Show how many movies remain under possible thresholds
for threshold in [1, 2, 5, 10, 20, 50, 100, 200, 500, 1000]:
    n_movies = movies_with_features_df.filter(col("num_ratings") >= threshold).count()
    print(f"Movies with at least {threshold} ratings: {n_movies:,}")


```

```python
from pyspark.sql.functions import col, lit, sum as spark_sum

# Use only movies with enough ratings for modelling.
# This reduces noise from movies where avg_rating is based on very few users.
MIN_RATINGS = 50
m = 50

model_df = movies_with_features_df.filter(col("num_ratings") >= MIN_RATINGS)

# Compute the global mean only from the filtered modelling data.
# Weight by num_ratings so this represents the average across individual ratings.
C = model_df.select(
    (spark_sum(col("avg_rating") * col("num_ratings")) / spark_sum(col("num_ratings"))).alias("global_mean")
).first()["global_mean"]

# Bayesian weighted average rating target.
model_df = model_df.withColumn(
    "weighted_avg_rating",
    (
        (col("num_ratings") / (col("num_ratings") + lit(m))) * col("avg_rating")
        +
        (lit(m) / (col("num_ratings") + lit(m))) * lit(C)
    )
)

print(f"Movies before filtering: {movies_with_features_df.count():,}")
print(f"Movies used for modelling: {model_df.count():,}")
print(f"Minimum ratings threshold: {MIN_RATINGS}")
print(f"Smoothing value m: {m}")
print(f"Global mean rating C: {C:.4f}")

model_df.select(
    "title",
    "avg_rating",
    "num_ratings",
    "weighted_avg_rating"
).orderBy(col("num_ratings").asc()).show(20, truncate=False)
```

From this point onward, the modelling dataset is `model_df`, which contains only movies with at least `MIN_RATINGS` ratings. The train/test split is applied after this filtering step, and the regression models use `weighted_avg_rating` as the prediction target instead of raw `avg_rating`.

# Feature Engineering for ML Regression

Our target variable is `weighted_avg_rating` — the average user rating each movie received after its release date.

### Features to transform:
- **genres**: Multi-label categorical → binary columns (one-hot) for each of the 19 genres
- **original_language**: Categorical → StringIndexer + OneHotEncoder (top languages kept, rest grouped as "other")
- **release_date**: Extract `release_year` and `release_month` as numeric features
- **runtime, cast_frequency, director_frequency**: Already numeric — keep as-is
- **num_ratings**: Excluded from features — it is a post-release outcome and would cause data leakage

### Columns dropped before modelling:
- `movieId`, `title`, `tmdb_id`: Identifiers, not predictive
- `genres` (original string), `release_date`, `original_language`: Replaced by their encoded versions


```python
from pyspark.sql.functions import year, month, col, split, explode, lit, when
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, StandardScaler
from pyspark.ml import Pipeline

# 1. Extract release_year and release_month from release_date
df = model_df \
    .withColumn("release_year", year(col("release_date"))) \
    .withColumn("release_month", month(col("release_date")))

# 2. Genre one-hot encoding (multi-label)
# Get all distinct genres
all_genres = (df
    .withColumn("genre", explode(split(col("genres"), "\\|")))
    .select("genre").distinct()
    .filter(col("genre") != "(no genres listed)")
    .rdd.flatMap(lambda x: x).collect()
)
all_genres.sort()
print(f"Genres found: {len(all_genres)} → {all_genres}")

# Create a binary column for each genre
for g in all_genres:
    df = df.withColumn(f"genre_{g}", when(col("genres").contains(g), 1).otherwise(0))

df.select("title", "genres", *[f"genre_{g}" for g in all_genres]).show(5, truncate=False)

# 3. Group rare languages as "other" (keep top 4)
from pyspark.sql.functions import count

top_languages = (df.groupBy("original_language")
    .agg(count("*").alias("cnt"))
    .orderBy(col("cnt").desc())
    .limit(4)
    .select("original_language")
    .rdd.flatMap(lambda x: x).collect()
)
print(f"Top languages: {top_languages}")

df = df.withColumn("language_grouped",
    when(col("original_language").isin(top_languages), col("original_language"))
    .otherwise("other"))


```

```python
# finding the cutoff for train/test split
from pyspark.sql.functions import percentile_approx, year

# Distribution of movies per year
df.withColumn("release_year", year(col("release_date"))) \
    .groupBy("release_year") \
    .count() \
    .orderBy("release_year") \
    .show(50)

# Find the 80th percentile date
print(f"Total movies: {df.count()}")

for cutoff in ["2017-01-01", "2018-01-01", "2019-01-01", "2020-01-01"]:
    train_count = df.filter(col("release_date") < cutoff).count()
    test_count = df.filter(col("release_date") >= cutoff).count()
    pct = train_count / df.count() * 100
    print(f"Cutoff {cutoff}: Train={train_count:,} ({pct:.1f}%) | Test={test_count:,} ({100-pct:.1f}%)")

```

```python
# Chronological split based on release_date
train_df = df.filter(col("release_date") < "2019-01-01")
test_df = df.filter(col("release_date") >= "2019-01-01")

print(f"Train: {train_df.count():,}  (before 2019)")
print(f"Test:  {test_df.count():,}  (2019 onwards)")

```

```python
#Pipeline

# StringIndexer: converts language strings to numeric indices
# (Spark's OneHotEncoder requires numeric input, not strings)
language_indexer = StringIndexer(inputCol="language_grouped", outputCol="language_index", handleInvalid="keep")

# OneHotEncoder: converts numeric index to binary vector
language_encoder = OneHotEncoder(inputCol="language_index", outputCol="language_vec")

# VectorAssembler: combines all features into a single vector column
numeric_cols = ["runtime", "release_year", "release_month",
                "cast_frequency", "director_frequency"]
genre_cols = [f"genre_{g}" for g in all_genres]

assembler = VectorAssembler(
    inputCols=numeric_cols + genre_cols + ["language_vec"],
    outputCol="features_raw",
    handleInvalid="skip"
)

# StandardScaler: normalizes each feature to mean=0, std=1
# (prevents features with large values like runtime=100 from dominating binary 0/1 features)
scaler = StandardScaler(inputCol="features_raw", outputCol="features", withMean=True, withStd=True)

# Build pipeline and fit on train
pipeline = Pipeline(stages=[language_indexer, language_encoder, assembler, scaler])
pipeline_model = pipeline.fit(train_df)

# Transform both sets using train fitted pipeline
train_final = pipeline_model.transform(train_df)
test_final = pipeline_model.transform(test_df)

print(f"Feature vector length: {train_final.select('features').first()[0].size}")
train_final.select("movieId", "title", "avg_rating", "weighted_avg_rating", "features").show(5, truncate=False)

```

# Models

## Linear Regression

Linear Regression creates a linear function y = w₁x₁ + w₂x₂ + ... + b,
trying to minimize the Mean Squared Error between actual and predicted values.
It is the simplest baseline regression model.


```python
from pyspark.ml.regression import LinearRegression
from pyspark.ml.evaluation import RegressionEvaluator

# Train Linear Regression model
lr = LinearRegression(
    featuresCol="features",
    labelCol="weighted_avg_rating",
    maxIter=100,
    regParam=0.01       # L2 regularization to prevent overfitting
)

lr_model = lr.fit(train_final)

# Predictions on test set
lr_predictions = lr_model.transform(test_final)
lr_predictions.select("title", "avg_rating", "weighted_avg_rating", "prediction").show(10, truncate=False)

# Evaluation
evaluator_rmse = RegressionEvaluator(labelCol="weighted_avg_rating", predictionCol="prediction", metricName="rmse")
evaluator_mae = RegressionEvaluator(labelCol="weighted_avg_rating", predictionCol="prediction", metricName="mae")
evaluator_r2 = RegressionEvaluator(labelCol="weighted_avg_rating", predictionCol="prediction", metricName="r2")

print(f"RMSE: {evaluator_rmse.evaluate(lr_predictions):.4f}")
print(f"MAE:  {evaluator_mae.evaluate(lr_predictions):.4f}")
print(f"R²:   {evaluator_r2.evaluate(lr_predictions):.4f}")

# Model coefficients
print(f"\nIntercept: {lr_model.intercept:.4f}")
print(f"Number of coefficients: {len(lr_model.coefficients)}")

```

```python
from pyspark.ml.feature import Tokenizer, StopWordsRemover, HashingTF, IDF

# ════════════════════════════════════════════════════════════════
# Add TF-IDF features from movie title
# Pipeline: Title → Tokenize → Remove Stopwords → TF → IDF
# (As covered in Lecture 10 & Lab 8)
# ════════════════════════════════════════════════════════════════

# Stage 1: Tokenize title into words
title_tokenizer = Tokenizer(inputCol="title", outputCol="title_words")

# Stage 2: Remove stopwords (e.g. "the", "a", "of")
title_remover = StopWordsRemover(inputCol="title_words", outputCol="title_filtered")

# Stage 3: HashingTF — convert words to term frequency vector
title_hashingTF = HashingTF(inputCol="title_filtered", outputCol="title_tf", numFeatures=256)

# Stage 4: IDF — weight by inverse document frequency
title_idf = IDF(inputCol="title_tf", outputCol="title_tfidf")

# ════════════════════════════════════════════════════════════════
# Updated pipeline with title TF-IDF features
# ════════════════════════════════════════════════════════════════

# New assembler: original features + title TF-IDF
assembler_v2 = VectorAssembler(
    inputCols=numeric_cols + genre_cols + ["language_vec", "title_tfidf"],
    outputCol="features_raw",
    handleInvalid="skip"
)

scaler_v2 = StandardScaler(inputCol="features_raw", outputCol="features", withMean=True, withStd=True)

# Full pipeline
pipeline_v2 = Pipeline(stages=[
    language_indexer, language_encoder,          # language encoding
    title_tokenizer, title_remover,             # title NLP
    title_hashingTF, title_idf,                 # title TF-IDF
    assembler_v2, scaler_v2                     # assemble + scale
])

# Fit on train, transform both
pipeline_model_v2 = pipeline_v2.fit(train_df)
train_v2 = pipeline_model_v2.transform(train_df)
test_v2 = pipeline_model_v2.transform(test_df)

print(f"New feature vector length: {train_v2.select('features').first()[0].size}")

# ════════════════════════════════════════════════════════════════
# Random Forest with title features
# ════════════════════════════════════════════════════════════════
rf_v2 = RandomForestRegressor(featuresCol="features", labelCol="weighted_avg_rating", numTrees=100, maxDepth=10, seed=42)
rf_model_v2 = rf_v2.fit(train_v2)
rf_preds_v2 = rf_model_v2.transform(test_v2)

print(f"\nRandom Forest WITH title TF-IDF:")
print(f"RMSE: {evaluator_rmse.evaluate(rf_preds_v2):.4f}")
print(f"MAE:  {evaluator_mae.evaluate(rf_preds_v2):.4f}")
print(f"R²:   {evaluator_r2.evaluate(rf_preds_v2):.4f}")

```
